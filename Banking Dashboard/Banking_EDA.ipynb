```python
bins = [0,100000,300000,float('inf')]
labels=['Low','Med','High']
df['Income Band']=pd.cut(df['Estimated Income'],bins=bins,labels=labels,right=False)
```

```python
df['Income Band'].value_counts().plot(kind='bar',color='orange')
```

```python
#Examine the Distribution of unique categories in categorical column
categorical_cols = df[["BRId","GenderId","IAId","Amount of Credit Cards","Nationality","Occupation","Fee Structure","Loyalty Classification","Properties Owned","Risk Weighting","Income Band"]].columns

for col in categorical_cols:
    print(f"Value Counts for '{col}':")
    display(df[col].value_counts())
```

##Univarient Analysis

```python
for i, predictor in enumerate(df[["BRId","GenderId","IAId","Amount of Credit Cards","Nationality","Occupation","Fee Structure","Loyalty Classification","Properties Owned","Risk Weighting","Income Band"]].columns):
     plt.figure(i, figsize=(3,3))
     plt.figure(i)
     sns.countplot(data=df,x=predictor)
     plt.show()
```

## Bivarient Analysis

```python
for i, predictor in enumerate(df[["BRId","GenderId","IAId","Amount of Credit Cards","Nationality","Fee Structure","Loyalty Classification","Properties Owned","Risk Weighting","Income Band"]].columns):
     plt.figure(i, figsize=(7,3))
     plt.figure(i)
     sns.countplot(data=df,x=predictor,hue = 'Nationality')
     plt.legend(fontsize = 9)
     sns.reset_defaults()
     plt.show()
```

```python
# HISTOGRAM of value counts for different occupation
for col in categorical_cols:
  if col == "Occupation":   # skip Occupation cols and move to next
    continue
  plt.figure(figsize=(7,4))
  sns.histplot(df[col])
  plt.title("Histogram of Occupation Count")
  plt.xlabel(col)
  plt.ylabel("Count")
  plt.show()
```

##Numerical analysis

```python
#Numerical analysis
numerical_cols = ['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Deposits','Checking Accounts','Foreign Currency Account','Business Lending']

#univariants analysis
plt.figure(figsize = (20,20))
for i, col in enumerate(numerical_cols):
  plt.subplot(4,3,i+1)
  sns.histplot(df[col],kde=True)
  plt.title(col)
plt.show()
```

## Heatmap

```python
numerical_cols = ['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Deposits','Checking Accounts','Foreign Currency Account','Business Lending']

correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(8, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5,fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

## Insights of EDA

1. The strongest positive correlation occur among "Bank Deposits" with "Checking Accounts", "Saving Accounts" and "Foreign Currency Account" indicating that customers who maintain high balances in one more account type often hold sustantial amount/funds across other aacounts as well
